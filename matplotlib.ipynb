# Matplotlib Library 📊

### Matplotlib is a popular Python library used for data visualization. It helps us represent data using different types of charts and graphs.

### The most commonly used module is pyplot.

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

## Line Chart

A line chart is used to show trends or changes in data over time.

plt.plot(x, y)

plt.show()

x → X-axis values,
y → Y-axis values,
plot() → creates the line chart,
show() → displays the chart

```python
df= pd.read_csv("../Stats & Prob/data files/sales_data_365_days.csv")
df
```

```python
# draw a line chart for this

plt.figure(figsize=(10,10))   # figure size of the plot
plt.plot(df['date'],
        df['sales'],
        color='red',
        marker='*',
        linestyle='--')
plt.title("My chart")
plt.xlabel('Date',fontsize=8)
plt.ylabel('Sales',fontsize=8)
plt.tight_layout()
plt.show()  #compulsary
```

```python
df['sales1']=df['sales'] + 10
```

```python
df
```

```python
plt.figure(figsize=(10,10))   # figure size of the plot
plt.plot(range(1,366),df['sales'])   # to do not overlap dates so range is used.

plt.title("My chart")
plt.xlabel('Day (1-365)',fontsize=8, fontweight='bold')
plt.ylabel('Sales',fontsize=8)

plt.xticks(range(1,366,30),rotation=45)    # change range of x axis, rotation-rotae the number on x axis
plt.yticks(range(100,250,30))

plt.tight_layout()
plt.show()  #compulsary
```

```python
# TWO CHARTS IN A SAME PLOT
```

```python
plt.figure(figsize=(10,10))   # figure size of the plot
plt.plot(range(1,366),df['sales'],label='Nashik Sales')   # to do not overlap dates so range is used.
plt.plot(range(1,366),df['sales1'],color='red', label='Pune Sales') 

plt.title("My chart")
plt.xlabel('Day (1-365)',fontsize=8, fontweight='bold')
plt.ylabel('Sales',fontsize=8)

plt.xticks(range(1,366,30),rotation=45)    # change range of x axis, rotation-rotae the number on x axis
plt.yticks(range(100,250,30))

plt.tight_layout()
plt.legend(loc=2)   # location means where to place legend box
plt.show()  #compulsary
```

## Bar Chart :

A bar chart is used to represent and compare different categories using rectangular bars. The length/height of each bar represents its value.

For example, comparing marks of students, sales of products, population of cities, etc.

```python
df=pd.read_excel("../Stats & Prob/data files/pivot.xlsx")
df
```

```python
s1=df.groupby('Party Name')['Total'].sum().sort_values(ascending=False)
```

```python
s1
```

```python
s1.index
```

```python
s1.values
```

```python
# Sales accoring to party Name

plt.figure(figsize=(14,17))   # figure size of the plot
bars=plt.bar(s1.index,s1.values)
plt.bar_label(bars, fmt="%.2f", padding =3, fontsize=9, fontweight='bold',color='red')
# pltbar_label= This is used to display the numerical value on top of each bar.

plt.title("My chart")
plt.xlabel('Party Name',fontsize=8, fontweight='bold')
plt.ylabel('Sales',fontsize=8)

plt.xticks(rotation=45)    # change range of x axis, rotation-rotae the number on x axis

plt.tight_layout()
plt.show()  #compulsary
```

```python
# Sales accoring to party Name  ## HORIZTONAL BAR

plt.figure(figsize=(14,17))   # figure size of the plot
bars=plt.barh(s1.index,s1.values)
plt.bar_label(bars, fmt="%.2f", padding =3, fontsize=9, fontweight='bold',color='red')
# pltbar_label= This is used to display the numerical value on top of each bar.

plt.title("My chart")
plt.xlabel('Party Name',fontsize=8, fontweight='bold')
plt.ylabel('Sales',fontsize=8)

plt.xticks(rotation=45)    # change range of x axis, rotation-rotae the number on x axis

plt.tight_layout()
plt.show()  #compulsary
```

## Histogram : 
A Histogram is a graph used to show the distribution of numerical data.

It divides the numerical data into intervals called bins and shows how many values fall into each interval.

```python
df=pd.read_csv("../Stats & Prob/data files/iris.csv")
df
```

```python
plt.hist(df['sepal_length'] , density=True)
plt.show()
```

## Scatter Plot  :
is used to show the relationship between two numerical variables.

It represents each observation as a point (dot) on the graph.

```python
plt.scatter(df['sepal_length'],df['sepal_width'], s=8,label='sepal') # change the size of the dots using s.
plt.show()
```

```python
plt.scatter(df['sepal_length'],df['sepal_width'], s=8,label='sepal') # change the size of the dots using s.
plt.scatter(df['petal_length'],df['petal_width'], s=8, color='r', label='petal')
plt.legend()
plt.show()
```

## Pie chart :
A Pie Chart is used to show the proportion or percentage of different categories in a whole.

The entire pie represents 100%, and each slice represents one category's contribution to the total.

```python
df=pd.read_excel("../Stats & Prob/data files/pivot.xlsx")
df
```

```python
# year wise sales
```

```python
r= pd.pivot_table(data=df,
              index=df['Date'].dt.year,
               values='Total').round(2)
r            
```

```python
r.index
```

```python
r['Total']
```

```python
plt.pie(r['Total'], labels= r.index, autopct='%.2f%%')
plt.legend()
plt.show()
```

## Subplot :
A subplot is used when we want to display multiple charts/plots inside a single figure.

For example, instead of creating one chart at a time, we can show:

- Bar chart
- Line chart
- Pie chart

in different sections of the same figure.

- plt.subplot(rows, columns, position)

```python
plt.figure(figsize=(14,7)) 

plt.subplot(2,2,1)
plt.pie(r['Total'], labels= r.index, autopct='%.2f%%')    # figure size of the plot
plt.legend()


plt.subplot(2,2,2)
bars=plt.barh(s1.index,s1.values)


plt.subplot(2,2,4)
plt.pie([2,9])

plt.subplot(2,2,3)
plt.pie([1,2])

plt.show() 
```
