# List vs Array
| Feature                 | Python List                                | NumPy Array                           |
| ----------------------- | ------------------------------------------ | ------------------------------------- |
| Library                 | Built into Python                          | Provided by NumPy                     |
| Data types              | Can store different data types             | Usually stores the same data type     |
| Speed                   | Relatively slower for numerical operations | Faster for numerical calculations     |
| Memory                  | Uses more memory                           | More memory-efficient                 |
| Mathematical operations | Need loops for many operations             | Supports direct vectorized operations |
| Dimensions              | Can represent nested lists                 | Naturally supports 1D, 2D, 3D, etc.   |
| Best use                | General-purpose data storage               | Numerical/scientific computing        |


list >> simple opertaions
array >> advance operations

array is FASTER than List
Extensible > array>>  c, python> speed

## Conditional slicing :
Conditional slicing means selecting elements from a NumPy array based on a condition rather than using fixed index positions.

It is commonly done using Boolean indexing.

```python
# eg
import numpy as np

a = np.array([2, 5, 8, 10, 3, 7])

result = a[a > 5]

print(result)
```

1. AND logical = &  -> returns true if both conditions are TRUE
