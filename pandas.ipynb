# nsmallest() and nlargest() in Pandas :
These functions are used to quickly find the smallest or largest n values from a DataFrame column.

1. nsmallest() : nsmallest() returns the n rows having the smallest values in a particular column.

Syntax
df.nsmallest(n, 'column_name')

# groupby / pivot table :

        a. sum()
        b. mean()

```python
df=pd.read_excel("../Stats & Prob/data files/pivot.xlsx")
df

```

```python
r1=df.groupby(by="Party Name")['Total'].sum()
```

```python
r1.sort_values(ascending=False)
```

### Pivot Table : 
    values=None,
    index=None,
    columns=None,
    aggfunc: 'AggFuncType' = 'mean',

```python
r4=pd.pivot_table(data=df,
               values='Total',
              index='Party Name',
                aggfunc='sum')

r4.round()
```

```python
r4.nlargest(3,'Total')
```

```python

r4=pd.pivot_table(data=df,
               values=['Total','Qty'],
              index='Party Name',
                aggfunc='sum')

r4.round()
```

```python
# drill down goes in detail of the thing
r4=pd.pivot_table(data=df,
               values=['Total'],
              index='Party Name',
                  columns='Sales Man',
                aggfunc='sum')

r4.round()
```

```python
# years wise sales
pd.pivot_table(
    data= df,
    values='Total',
    index=df['Date'].dt.year,
    aggfunc=['sum']
).round()
```

```python
# years wise sales using multiplle aggfunctsions

pd.pivot_table(
    data= df,
    values='Total',
    index=df['Date'].dt.year,
    aggfunc=['sum','mean','median']
).round()
```

```python
# 3 items sales are high

pd.pivot_table(data=df,
              values='Total',
              index='Item Name',
               aggfunc='sum'
              ).nlargest(3,'Total')

```

# apply / map 

1. apply() : apply() is used to apply a function to each value, row, or column.
2. map() : map() is mainly used with a Series (single column) to transform or replace values.

```python
df=pd.read_csv("../Stats & Prob/data files/students.csv")
df
```

```python
# apply column divide by 2

df['sq of percantege'] = df['Percantege'].apply(lambda x : x**2)
df
```

```python
# length of a percentage column

df['Len of %'] = df['Percantege'].apply(lambda x : len(str(x)))
```

```python
df
```

```python
df['Grade'] = df['Percantege'].apply(lambda x : 'Good' if x> 70 else 'Bad')
```

```python
df
```

```python
# map : apply functions on column or Series 
# also used for replcaing values
```

```python
df['Grade'] = df['Percantege'].map(lambda x : 'Good' if x> 70 else 'Bad')
```
