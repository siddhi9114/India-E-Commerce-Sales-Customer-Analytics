# Seaborn : 
Seaborn is a Python library used for data visualization. It is built on top of Matplotlib and provides attractive, easy-to-create statistical graphs.

 | Seaborn Function    | Used For                                   |
| ------------------- | ------------------------------------------ |
| `sns.histplot()`    | Distribution of data                       |
| `sns.countplot()`   | Count of categories                        |
| `sns.barplot()`     | Compare numerical values across categories |
| `sns.boxplot()`     | Distribution and outliers                  |
| `sns.scatterplot()` | Relationship between two variables         |
| `sns.lineplot()`    | Trend over time                            |
| `sns.heatmap()`     | Correlation/matrix visualization           |
| `sns.pairplot()`    | Relationships between many variables       |
| `sns.violinplot()`  | Distribution + density                     |


```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
sns.get_dataset_names()


# some data set are presesnt in seaborn which we can use 
```

```python
df= sns.load_dataset('tips')    # to load dataset named tips
df
```

# Seaborn Line Chart 📈

A Seaborn line chart is used to show trends or changes in numerical data, especially over an ordered variable such as time.

Seaborn provides the sns.lineplot() function for creating line charts.

```python
# total bill wise tips

sns.lineplot(data=df,
            x='total_bill',
            y='tip')

plt.show()
```

```python
sns.lineplot(data=df,
            x='total_bill',
            y='tip')
plt.title('total bill wise tips')
plt.show()
```

```python
# if data=df is not mentin write in list otherwise in strings is also okay!!

sns.lineplot(x=df['total_bill'],
            y=df['tip'])
plt.title('total bill wise tips')
plt.show()
```

### hue parameter :  is used to divide data into different groups/categories and represent each group separately in a chart.

```python

sns.lineplot(data=df,
            x='total_bill',
            y='tip',
            hue='sex')   # hue sepeearte as per sex of the person
plt.title('total bill wise tips')
plt.show()
```

```python
sns.lineplot(data=df,
            x='total_bill',
            y='tip',
            hue='time')   # hue sepeearte as per time
plt.title('total bill wise tips')
plt.show()
```

# Seaborn Bar plot : barplot() is a Seaborn function used to compare a numerical value across different categories.

```python
plt.figure(figsize=(6,7))
ax = sns.barplot(data=df,
           x='time',
           y='total_bill',
           estimator='sum',   # automatically aggregates no need of pivot table or gtoupby
            errorbar=None)  # to remove error bar from the output
                 
# data labels
for i in ax.containers:
    ax.bar_label(i,fmt='%.2f', padding=2, fontweight='bold')

plt.title("time and total bill")
plt.xlabel('Time')
plt.show()
```

## Scatter Plot in Seaborn 📊

A scatter plot in Seaborn is used to show the relationship between two numerical variables. Each observation is represented by a dot.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv('../Stats & Prob/data files/iris.csv')
df
```

```python
sns.scatterplot(data=df,
               x='sepal_length',
               y='sepal_width',
               hue='species')  # 

plt.show()
```

# Histogram in Seaborn 📊

A histogram is used to show the distribution of numerical data. It tells us how many observations fall within different ranges (bins).

```python
sns.histplot(data=df,
            x='sepal_length',
            kde=True)  # adds a smooth curve showing the approximate distribution of the data.
plt.show()
```

# Seaborn Boxplot

A boxplot is used to show the distribution of numerical data and identify outliers.

```python
sns.boxplot(data=df,x='sepal_length',
           hue='species')   # wants category wise

plt.show()
```

# Seaborn pairplot()

combination of two charts >> histogram and scatterplot

A pairplot is used to visualize relationships between multiple numerical columns at the same time.

```python
sns.pairplot(data=df)
```

# Seaborn heatmap()
- shows 2D data, dataframe
- A heatmap is used to represent data values in a matrix using different colors.
- It is commonly used to show a correlation matrix between numerical variables.

```python
cm=df.corr(numeric_only=True)
cm
```

```python
# correlation shows this

sns.heatmap(cm,
            annot=True)  # shows values in written
```

# Plotly

Plotly is a Python library used to create interactive and visually attractive data visualizations.

Unlike Matplotlib and Seaborn, Plotly allows you to hover, zoom, pan, and interact with the chart.

### 1. Matplotlib → Basic & flexible
### 2. Seaborn → Statistical & beautiful
### 3. Plotly → Interactive

```python
import plotly.express as px
fig=px.scatter(df, x= 'sepal_length',y='sepal_width', color='species')
fig
```
